# Task 2 - Conditional Statements

```python
def get_input():
    number = input('Enter a number: ')
    try:
        number = float(number)
        return number
    except ValueError:
        print('Invalid input. Please enter a valid number.')
        return None

def check_positive_negative_zero(number):
    if number > 0:
        return 'positive'
    elif number < 0:
        return 'negative'
    else:
        return 'zero'

def main():
    number = get_input()
    if number is not None:
        result = check_positive_negative_zero(number)
        print('The number is {}' .format(result))

if __name__ == "__main__":
    main()
```

# Task 3 - Loops and Iteration

```python
def get_input():
    nnn_number = input('Enter a positive integer bigger than zero: ')
    try:
        nnn_number = int(nnn_number)
        if nnn_number <= 0:
            raise ValueError('The number must be positive and bigger than zero.')
        return nnn_number
    except ValueError:
        print('Invalid input. Please enter a valid positive integer bigger than zero.')
        return None

def is_even(number):
    return number % 2 == 0

def get_even_number(max_number):
    list_of_even = []
    for i in range(1, max_number + 1):
        if is_even(i):
            list_of_even.append(i)
    return list_of_even

def main():
    nnn_number = get_input()
    if nnn_number is not None:
        even_number = get_even_number(nnn_number)
        if even_number:
            print('The even numbers from 1 to {} are: {}' .format(nnn_number, ', '.join(map(str, even_number))))
        else:
            print('There are no even numbers in the range from 1 to {}.' .format(nnn_number))

if __name__ == "__main__":
    main()
```

# Task 4 - Functions

```python
def calculate_rectangle_area(length, width):
    return length * width

def get_input():
    length = input('Enter the length of the rectangle: ')
    width = input('Enter the width of the rectangle: ')
    try:
        length = float(length)
        width = float(width)
        if length <= 0 or width <= 0:
            raise ValueError('Length and width must be positive.')
        return length, width
    except ValueError:
        print('Invalid input. Please enter valid positive numbers.')
        return None, None

def main():
    length, width = get_input()
    if length is not None and width is not None:
        area = calculate_rectangle_area(length, width)
        print('The area of the rectangle is: {}' .format(area))

if __name__ == "__main__":
    main()
```
