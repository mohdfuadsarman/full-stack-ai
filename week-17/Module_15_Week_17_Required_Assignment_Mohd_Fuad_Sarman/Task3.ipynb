# Newswire Classification using ANN

```python
import numpy as np
from keras.datasets import reuters
from keras.preprocessing.sequence import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, LSTM, Dense
from keras.utils import to_categorical
```

a) Load and preprocess the Reuters dataset using keras.datasets.reuters

```python
num_words = 10000  # Only keep top 10,000 most frequent words (vocabulary size)

(x_train, y_train), (x_test, y_test) = reuters.load_data(num_words=num_words)

print(f"Number of training samples: {len(x_train)}")
print(f"Number of test samples: {len(x_test)}")
print(f"\nExample of a tokenized news article (train[0]):\n{x_train[0]}")
print(f"\nCorresponding category (label): {y_train[0]}")
print(f"\nNumber of unique categories: {len(np.unique(y_train))}")
```

b) Apply padding to ensure all input sequences are of the same length

```python
maxlen = 200  # Pad sequences to 200 words

x_train = pad_sequences(x_train, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)

print(f"Padded train sample (x_train[0]):\n{x_train[0]}")
print(f"\nShape of x_train: {x_train.shape}")
```

c) One-hot encode the target labels for multi-class classification

```python
num_classes = np.max(y_train) + 1  # Should be 46 output classes

print(f"Number of classes: {num_classes}")

y_train = to_categorical(y_train, num_classes)
y_test = to_categorical(y_test, num_classes)

print(f"\nEncoded label (y_train[0]):\n{y_train[0]}")
print(f"\nShape of y_train: {y_train.shape}")
```

d) Build an LSTM-based model using Keras; train, validate, and evaluate the model on unseen test data.

```python
print("Building the model...")

model = Sequential()
model.add(Embedding(input_dim=num_words, output_dim=128, input_length=maxlen)) # an Embedding layer with input_dim=10000, output_dim=128 and input_length=200
model.add(LSTM(64)) # one LSTM layer with 64 units
model.add(Dense(num_classes, activation='softmax')) # a Dense output layer with activation='softmax' for 46 output classes
```

### Model summary

```python
print("Model summary:")

model.summary()
```

### Compile and train the model

```python
print("Compiling the model...")

model.compile(optimizer='adam',  # Optimizer for gradient descent (Adam)
              loss='binary_crossentropy',  # Loss function for binary classification
              metrics=['accuracy'])  # Metric to monitor during training (accuracy)

print("\nTraining the model...")

batch_size = 32 # Small batch size for training will help with convergence, big batch size will help with speed
epochs = 10

history = model.fit(x_train, y_train,  # Training data and labels
                    batch_size=batch_size,  # Number of samples per gradient update
                    epochs=epochs,  # Number of iterations over the entire dataset
                    validation_data=(x_test, y_test))  # Data on which to evaluate the loss and any model metrics at the end of each epoch
```

### Evaluation

```python
print("Evaluating the model...")
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"\nTest Accuracy: {test_acc * 100:.2f}%")
```
