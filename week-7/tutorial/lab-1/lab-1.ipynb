# Synthetic Data Generation

```python
import numpy as np
import time

np.random.seed(int(round(time.time())))

# Simulate 5000 customers' purchase amounts (in dollars)
# purchase data will follow a normal distribution with mean $100 and standard deviation $25
purchase_amount = np.random.normal(loc=100, scale=25, size=5000)
# Clip to ensure that there is no negative spending
purchase_amount = np.clip(purchase_amount, 0, None)
print(purchase_amount)
```

# Basic Data Analysis

```python
# Find average purchase amount
average_purchase = np.mean(purchase_amount)
print(f"Average purchase amount: ${average_purchase:.2f}")
```

```python
# Find standard deviation of purchase amounts
std_dev_purchase = np.std(purchase_amount)
print(f"Standard deviation of purchase amounts: ${std_dev_purchase:.2f}")
```

```python
# Find purchase amounts above $100
above_100 = purchase_amount[purchase_amount > 100]
print(f"Number of purchases above $100: {len(above_100)}")
print(f"Percentage of purchases above $100: {len(above_100) / len(purchase_amount) * 100:.2f}%")
```

```python
# Find top 5% of purchases
top_95_percent = np.percentile(purchase_amount, 95)
top_5_percent_spenders = purchase_amount[purchase_amount >= top_95_percent]
print(f"Top 5% threshold: ${top_95_percent:.2f}")
print(f"Number of top 5% purchases: {len(top_5_percent_spenders)}")
```

```python
# Calculate total revenue
total_revenue = np.sum(purchase_amount)
print(f"Total revenue: ${total_revenue:.2f}")
```

# Advance Data Analysis

```python
# Get outliers (purchases above 2 standard deviations from the mean)
mean_purchase = np.mean(purchase_amount)
std_dev_purchase = np.std(purchase_amount)
outliers_threshold_high = mean_purchase + 2 * std_dev_purchase
outliers_threshold_low = mean_purchase - 2 * std_dev_purchase
outliers_high = purchase_amount[purchase_amount > outliers_threshold_high]
outliers_low = purchase_amount[purchase_amount < outliers_threshold_low]
print(f"Outliers threshold (high): {outliers_threshold_high}")
print(f"Number of high outliers: {len(outliers_high)}")
print(f"Outliers threshold (low): {outliers_threshold_low}")
print(f"Number of low outliers: {len(outliers_low)}")
```

```python
# Group customers by purchase amount ranges
purchase_bins = [0, 50, 100, 150, 175]
purchase_labels = ['< $50', '$50 - $100', '$100 - $150', '$150 - $175','> $175']
purchase_categories = np.digitize(purchase_amount, purchase_bins)
print("Purchase categories:")
for i, label in enumerate(purchase_labels):
    count = np.sum(purchase_categories == i + 1)
    print(f"{label}: {count} purchases")
```

```python
# Simulate price increase
price_increase = 1.15  # 15% price increase
purchase_amount_increased = purchase_amount * price_increase

projected_revenue = np.sum(purchase_amount_increased)
print(f"Projected revenue after 15% price increase: ${projected_revenue:.2f}")
print(f"Projected revenue increase: ${projected_revenue - total_revenue:.2f}")
```
