# Customer Purchase Data Analysis with NumPy

```python
import numpy as np
import time
```

## Exercise 1 - Synthetic Data Generation

```python
np.random.seed(int(round(time.time())))
```

```python
# Simulate 5000 customers' purchase amounts (in dollars)
# purchase data will follow a normal distribution with mean $100 and standard deviation $25
purchase_amounts = np.random.normal(loc=100, scale=25, size=5000)
# Clip to ensure that there is no negative spending
purchase_amounts = np.clip(purchase_amounts, 0, None)
purchase_amounts
```

# Exercise 2 - Basic Data Analysis

```python
# Average spending
average_spending = np.mean(purchase_amounts)
print('Average spending: ${:.2f}'.format(average_spending))
```

```python
# Standard deviation of spending
std_spending = np.std(purchase_amounts)
print('Spending standard deviation: ${:.2f}'.format(std_spending))
```

```python
# Customers who spent more than $100
high_value_customers = purchase_amounts[purchase_amounts > 100]

print('Number of high-value customers: {}'.format(len(high_value_customers)))
print('Percentage of high-value customers: {:.2f}%'.format((len(high_value_customers) / len(purchase_amounts)) * 100))
```

```python
# 95th percentile spending threshold
top_5_percent_threshold = np.percentile(purchase_amounts, 95)

# Find customers who are in the top 5%
top_spenders = purchase_amounts[purchase_amounts >= top_5_percent_threshold]

print('Number of top spenders: {}'.format(len(top_spenders)))
print('Top 5% spending threshold: ${:.2f}'.format(top_5_percent_threshold))
```

```python
total_revenue = np.sum(purchase_amounts)
print('Total revenue from the sale event: ${:.2f}'.format(total_revenue))
```

# Exercise 3 - Advanced Data Analysis

```python
# Define outliers as customers who spent more than mean +/= 2*std
mean = np.mean(purchase_amounts)
std = np.std(purchase_amounts)

outlier_threshold_high = mean + 2 * std
outlier_threshold_low = mean - 2 * std
outliers_high = purchase_amounts[purchase_amounts > outlier_threshold_high]
outlier_low = purchase_amounts[purchase_amounts < outlier_threshold_low]

print('Outlier high spending threshold: ${:.2f}'.format(outlier_threshold_high))
print('Outlier low spending threshold: ${:.2f}'.format(outlier_threshold_low))

print('Number of high outliers: {}'.format(len(outliers_high)))
print('Number of low outliers: {}'.format(len(outlier_low)))
```

```python
# Define bins
bins = [0, 50, 100, 150, 175]
labels = ['Very Low Spender', 'Low Spender', 'Average Spender', 'High Spender', 'Very High Spender']

# Digitize spending
spending_tiers = np.digitize(purchase_amounts, bins, right=True)

# Count customers in each tier
for i, label in enumerate(labels):
    count = np.sum(spending_tiers == i+1)
    print(f"{label}: {count} customers")
```

```python
# Simulate new purchase amounts after marketing campaign
expected_increase_rate = 1.15
new_purchase_amounts = purchase_amounts * expected_increase_rate

new_total_revenue = np.sum(new_purchase_amounts)

print('Projected total revenue after campaign: ${:.2f}'.format(new_total_revenue))
print('Projected revenue increase: ${:.2f}'.format(new_total_revenue - total_revenue))
```
