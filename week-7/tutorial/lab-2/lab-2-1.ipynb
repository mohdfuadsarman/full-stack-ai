# Basic Pandas Operation

```python
# a import necessary library
import pandas as pd
```

```python
# b import data.csv
# set user_id as row index
data = pd.read_csv('data/occupation.csv', index_col=0)
```

```python
# c print the first 25 rows
data.head(25)
# data.iloc[:25]
```

```python
# d print the last 10 rows
data.tail(10)
# data.iloc[-10:]
```

```python
# e print number of observations
print('Number of observations with len = ', len(data))
print('Number of observations with shape = ', data.shape[0])
print('Number of observations with index = ', data.index.size)
```

```python
# f print number of columns
print('Number of columns with len = ', len(data.columns))
print('Number of columns with shape = ', data.shape[1])
print('Number of columns with index = ', data.columns.size)
```

```python
# g print the column names
# data.columns
print('Column names = ', data.columns.tolist())
```

```python
# h print the index numbers
data.index
```

```python
# i print data types of each column
data.dtypes
```

```python
# j print occupation column
# data['occupation']
data.occupation
```

```python
# k count number of occupations
print('Number of occupations = ', data.occupation.unique().size)
print('Number of occupations = ', data.occupation.nunique())
```

```python
# l print list of user aged 50 and above
print('Users aged 50 and above:')
print(data[data.age >= 50].index.tolist())

# count number of users aged 50 and above
print('Number of users aged 50 and above = ', data[data.age >= 50].shape[0])
```

```python
# m print descriptive statistics
print('Descriptive statistics:')
data.describe()
```

```python
# n print descriptive statistic for all columns
# set include='all' to include non-numeric columns, otherwise it will only include numeric columns
print('Descriptive statistics for all columns:')
data.describe(include='all')
```

```python
# o print descriptive statistics for occupation column
print('Descriptive statistics for occupation column:')
data.occupation.describe()
```

```python
# p calculate mean age
mean_age = data.age.mean()
print('Mean age = ', mean_age)
```

```python
# q print age with least occurrences
data.age.value_counts().tail()
```

```python
# r add new salary column with initial value of 0
data['salary'] = 0
data
```

```python
# s set salary of each user to 100 multiplied by their age
data['salary'] = data.age * 100
data
```
